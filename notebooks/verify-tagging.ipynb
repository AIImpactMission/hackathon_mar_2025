```python
import os
import json

def load_sample_output(batch_idx=0, folder="batch_results"):
    """
    Loads a sample output JSON file from the specified folder.
    
    Args:
        batch_idx (int): Index of the batch file to load.
        folder (str): Directory where batch result files are stored.
        
    Returns:
        dict or None: Parsed JSON data if successful, otherwise None.
    """
    sample_file = os.path.join(folder, f"batch_{batch_idx}.json")
    try:
        with open(sample_file, "r", encoding="utf-8") as f:
            data = json.load(f)
        return data
    except Exception as e:
        print(f"Error loading file {sample_file}: {e}")
        return None

DATA_DIR = "/export/share/krausef99dm/AIM_delete/data/2025_02_25_wienerzeitung_archiv/batch_results"

sample_data = load_sample_output(400, folder=DATA_DIR)
if sample_data is not None:
    print("Sample output:")
    print(json.dumps(sample_data, indent=2))
```
